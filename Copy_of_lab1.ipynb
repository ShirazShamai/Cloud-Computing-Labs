```python
# @title Default title text
name = "Shiraz Shamai" # @param ["Shiraz Shamai","Yael Wroblevsky"]
import ipywidgets as widgets
from IPython.display import display
button = widgets.Button(description="Click Me!")
output = widgets.Output()

style = {'description_width': '120px'}
layout = widgets.Layout(width='750px')
student_name = widgets.Text(description="Name:", style=style, layout=layout)
student_email = widgets.Text(description="Email:", style=style, layout=layout)
student_courses = widgets.Text(description="Courses:", style=style, layout=layout)
student_link = widgets.Text(description="Link:", style=style, layout=layout)
favorite_show = widgets.Text(description="Favorite Show:", style=style, layout=layout)

with open('/content/drive/My Drive/מחשוב ענן/students.txt', 'r') as file:
    contents = file.read()

for line in contents.splitlines():
    parts = line.split(";")

    if len(parts) >= 4 and parts[0].strip() == name:
        student_name.value = parts[0].strip()
        student_email.value = parts[1].strip()
        student_courses.value = parts[2].strip()
        student_link.value = parts[3].strip()
        break

update_button = widgets.Button(description="Update Show")

def update_show(b):
    with open('/content/drive/My Drive/מחשוב ענן/students.txt', 'a') as file:
        file.write("\n" + name + ";" + favorite_show.value)

    print("Favorite show updated successfully!")

favorite_show.value = ""

for line in contents.splitlines():
    parts = line.split(";")

    if len(parts) == 2 and parts[0].strip() == name:
        favorite_show.value = parts[1].strip()

update_button.on_click(update_show)

display(
    student_name,
    student_email,
    student_courses,
    student_link,
    favorite_show,
    update_button
)

```
