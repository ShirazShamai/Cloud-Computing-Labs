```python
import json

students = [
    {
        "first_name": "Shiraz",
        "last_name": "Shamai",
        "email": "shiraz.shamai@e.braude.ac.il",
        "courses": ["Cloud Computing", "WEB", "Intro to AI", "Data Mining", "Compiler Theory", "Ethics & AI"],
        "link": "https://www.google.com"
    },
    {
        "first_name": "Yael",
        "last_name": "Wroblevsky",
        "email": "Yael.Wroblevsky@e.braude.ac.il",
        "courses": ["Cloud Computing", "WEB", "Probability", "Algorithms"],
        "link": "https://w3.braude.ac.il/"
    }
]
```

```python
from google.colab import drive
drive.mount('/content/drive')

with open('/content/drive/My Drive/מחשוב ענן/lab2/students.json', 'w') as file:
    json.dump(students, file, indent=4)
```

```python
with open('/content/drive/My Drive/מחשוב ענן/lab2/students.json', 'r') as file:
    students = json.load(file)

print(students)
```

```python
# @title Default title text
selected_student = "Shiraz Shamai" # @param ["Shiraz Shamai","Yael Wroblevsky"]

for student in students:
    full_name = student["first_name"] + " " + student["last_name"]

    if full_name == selected_student:
      first_name = student["first_name"]
      last_name = student["last_name"]
      email = student["email"]
      courses = ", ".join(student["courses"])
      link = student["link"]
      favorite_show = student.get("favorite_show", "")

import ipywidgets as widgets
from IPython.display import display

first_name_box = widgets.Text(value=first_name, description="First Name:")
last_name_box = widgets.Text(value=last_name, description="Last Name:")
email_box = widgets.Text(value=email, description="Email:")
courses_box = widgets.Text(value=courses, description="Courses:")
link_box = widgets.Text(value=link, description="Link:")
favorite_show_box = widgets.Text(
    value=favorite_show or "",
    description="Favorite Show:"
)
button = widgets.Button(description="Update")
display(button)

def update_student(b):
    for student in students:
        full_name = student["first_name"] + " " + student["last_name"]

        if full_name == selected_student:
            student.update({"favorite_show": favorite_show_box.value})
            favorite_show_box.value = student["favorite_show"]

    with open('/content/drive/My Drive/מחשוב ענן/lab2/students.json', 'w') as file:
        json.dump(students, file, indent=4)

    print("Updated successfully!")

button.on_click(update_student)

display(first_name_box, last_name_box, email_box, courses_box, link_box, favorite_show_box)
```
